```python
# IMPORTANT: Due to Windows dynamic library / runtime state collision torch must be imported before mlrun (local running)
import torch
import mlrun

mlrun.set_environment(api_path="http://localhost:30070")

project = mlrun.load_project(
    name="legalcontractextractor", context="../"
)  # project yaml must be in this directory
# Verify it loaded correctly by checking its status or printing the config
# print(project.to_yaml())
```

```python
"""
UIDs:
f2eb59a0303e47449661a5db7f18805a 1st
615a85ee5f154264a6042d2f2b744b21 2nd
"""
UID = "f2eb59a0303e47449661a5db7f18805a"
model_key = "Hermes4-14b-ContractModel"

run_obj = project.run_function(
    function="register-model",    # use the name in the register.ipynb file
    params={
        "experiment_run_uid":UID,
        "model_key": model_key,
        "project_name": project.name
    },
    local=True,
    watch=True,
    verbose=True
)
```

```python
import mlrun
from mlrun.model import RunObject

UID = "615a85ee5f154264a6042d2f2b744b21"

# Initialize the MLRun DB client
db = mlrun.get_run_db()
run_dict = db.read_run(uid=UID, project="legalcontractextractor")
# Convert the dictionary to a RunObject for easier API access
run = RunObject.from_dict(run_dict)

# Get run outputs
run_parameters = run_dict['spec']['parameters']
run_metrics = run_dict['status']['results']
output = run.outputs['return'] # this is what was returned 

print(run_parameters)
print(run_metrics)
print(output)
```
