## Data registration

```python
# https://docs.mlrun.org/en/v1.10/api/mlrun.projects/index.html#mlrun.projects.MlrunProject.set_function

data_reg_func = project.set_function(
    func="src/register_raw_dataset.py",
    name="raw-proc",
    kind="job",
    image="mlrun/mlrun",
    handler="process_raw",
)
```
