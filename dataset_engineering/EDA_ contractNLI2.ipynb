```python
x = df_short[df_short['document_id'] == 86][['hypothesis_id', 'hypothesis', 'label', 'evidence_texts']]
x.rename(columns={"evidence_texts":"source_clause"}, inplace=True)

def join_text(row):
    joined = ". ".join(row['source_clause'])
    return joined

x['source_clause'] = x.apply(join_text, axis=1)
x.to_dict(orient="records")
```
